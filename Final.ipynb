```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

numerical_columns = ['Previous Purchases', 'Purchase Amount (USD)', 'Review Rating']

features = data[numerical_columns]

# Standardize the numerical features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Use the Elbow Method to find the optimal number of clusters (K)
k_values = range(1, 11)

inertia_values = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(features_scaled)
    inertia_values.append(kmeans.inertia_)

plt.plot(k_values, inertia_values, marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Sum of Squared Distances (Inertia)')
plt.show()
```

```python
optimal_k = 4
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans.fit(features_scaled)
data['Cluster'] = kmeans.labels_
centroids = kmeans.cluster_centers_
plt.scatter(features_scaled[:, 0], features_scaled[:, 1], c=data['Cluster'], cmap='viridis', alpha=0.7, label='Data Points')


plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200, c='red', label='Centroids')

plt.title('K-means Clustering')
plt.xlabel('Previous Purchases')
plt.ylabel('Purchase Amount (USD)')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_avg = silhouette_score(features_scaled, data['Cluster'])
print(f"Silhouette Score: {silhouette_avg}")
```

```python
from sklearn.metrics import davies_bouldin_score
kmeans_labels = kmeans.fit_predict(features_scaled)
db_index = davies_bouldin_score(features_scaled, data['Cluster'])
print(f"Davies-Bouldin Index: {db_index}")
```

```python
from sklearn.metrics import calinski_harabasz_score
kmeans_labels = kmeans.fit_predict(features_scaled)
ch_index = calinski_harabasz_score(features_scaled, data['Cluster'])
print(f"Calinski-Harabasz Index: {ch_index}")
```

```python
#silhouette score indicates there is some separation between clusters but not much
#davies-bouldin index indicates that the quality of the clusters are not bad but not great
#calinski-harabasz index of 1451.08 is relatively high, suggesting that the clusters are well-separated.
```

```python
data
df_cleaned = data[['Frequency of Purchases','Subscription Status', 'Previous Purchases', 'Review Rating']]
df_cleaned
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
# Creating a LabelEncoder instance to encode categorical columns
label_encoder = LabelEncoder()
categorical_columns = ['Subscription Status','Frequency of Purchases']

for col in categorical_columns:
    df_cleaned[col] = label_encoder.fit_transform(df_cleaned[col])
# Separating features (X) and target variable (y)
X = df_cleaned.drop(['Subscription Status'], axis=1)
y = df_cleaned['Subscription Status']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
# Creating a Decision Tree Classifier model
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)

print(f"Accuracy: {accuracy}")
print("Confusion Matrix:")
print(conf_matrix)
```

```python
#The model's overall accuracy is ok (62.56%), but precision and recall for the positive class are low. 
#This suggests that while the model is reasonably accurate 
#in predicting instances without subscriptions (True Negative) 
#it struggles to correctly identify instances with subscriptions (True Positive)
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print(f"Precision: {precision}")
print(f"Recall: {recall}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

y_pred_probs = model.predict_proba(X_test)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, y_pred_probs)

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 8))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Guess')
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR) / Recall')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

```python
#It shows that the decision tree is about the same as randomly guessing
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


label_encoder = LabelEncoder()
categorical_columns = ['Subscription Status','Frequency of Purchases']

for col in categorical_columns:
    df_cleaned[col] = label_encoder.fit_transform(df_cleaned[col])

X = df_cleaned.drop(['Subscription Status'], axis=1)
y = df_cleaned['Subscription Status']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
# Defining the hyperparameter grid for Random Forest
param_grid = {
    'n_estimators': [50, 100, 150, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf_classifier = RandomForestClassifier(random_state=42)
# Performing Randomized Search for hyperparameter tuning
random_search = RandomizedSearchCV(
    rf_classifier,
    param_distributions=param_grid,
    n_iter=10,  
    scoring='accuracy', 
    cv=5,  
    random_state=42
)
# Fitting the model with Randomized Search to find the best hyperparameters
random_search.fit(X_train, y_train)
# Retrieving the best hyperparameters from Randomized Search
best_params = random_search.best_params_

y_pred = random_search.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
classification_rep = classification_report(y_test, y_pred)

print(f"Best Parameters: {best_params}")
print(f"Accuracy: {accuracy}")
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", classification_rep)
```

```python
#the best parameters are n_estimators=100, min_samples_split=5, min_samples_leaf=4, and max_depth=10.
#The confusion matrix displays the number of True Positive (279), True Negative (1), False Positive (3), and False Negative (107) predictions.
#the model has high accuracy due to a high number of True Negatives. 
#However, the recall for class 1(positive class) is low, suggesting that the model struggles to identify instances of class 1. 
#Further reflected in the low precision, recall, and F1-score for class 1 in the classification report.
```
